```python
import pandas as pd
import numpy as np

csv_path = "/content/LCL_Data/CC_LCL-FullData.csv"

df = pd.read_csv(csv_path, nrows=500000)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
```

```python
df['DateTime'] = pd.to_datetime(
    df['DateTime'], errors='coerce'
)

df['KWH/hh (per half hour)'] = pd.to_numeric(
    df['KWH/hh (per half hour)'],
    errors='coerce'
)

df = df.dropna(
    subset=['LCLid', 'DateTime', 'KWH/hh (per half hour)']
)

df['stdorToU'] = df['stdorToU'].fillna('Unknown')

print("Missing Values Handled!")
```

```python

import pandas as pd
import numpy as np

df = pd.read_csv('/content/cleaned_electricity.csv')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

# =====================================
# DATA CLEANING AND CORRECTIONS
# =====================================

energy = 'KWH/hh (per half hour)'

# 1. Remove extra spaces from text columns
for col in ['LCLid', 'stdorToU', 'Day_Name',
            'Consumption_Category']:
    df[col] = df[col].astype('string').str.strip()

# 2. Correct data types
df['DateTime'] = pd.to_datetime(
    df['DateTime'], errors='coerce'
)

df[energy] = pd.to_numeric(
    df[energy], errors='coerce'
)

# 3. Remove missing essential values
df = df.dropna(
    subset=['LCLid', 'stdorToU', 'DateTime', energy]
).copy()

# 4. Remove infinite and negative readings
df = df[np.isfinite(df[energy])].copy()
df = df[df[energy] >= 0].copy()

# 5. Remove duplicate records
df = df.drop_duplicates()

df = df.drop_duplicates(
    subset=['LCLid', 'DateTime'],
    keep='first'
)

# 6. Keep valid half-hour timestamps only
valid_time = (
    df['DateTime'].dt.minute.isin([0, 30]) &
    (df['DateTime'].dt.second == 0)
)

df = df[valid_time].copy()

# 7. Recreate date-related columns
df['Hour'] = df['DateTime'].dt.hour
df['Year'] = df['DateTime'].dt.year
df['Month'] = df['DateTime'].dt.month
df['Day'] = df['DateTime'].dt.day
df['Day_Name'] = df['DateTime'].dt.day_name()

# 8. Recalculate consumption category
df['Consumption_Category'] = np.select(
    [
        df[energy] <= 0.5,
        df[energy] <= 1.0
    ],
    ['Low', 'Medium'],
    default='High'
)

# 9. Sort records
df = df.sort_values(
    ['LCLid', 'DateTime']
).reset_index(drop=True)

print("Data corrections completed!")
print("Final shape:", df.shape)
```

```python

print("========== FINAL DATA QUALITY CHECK ==========")

print("\n1. Missing values:")
print(df.isnull().sum())

print("\n2. Duplicate records:")
print(df.duplicated().sum())

print("\n3. Negative consumption:")
print((df['KWH/hh (per half hour)'] < 0).sum())

print("\n4. Zero consumption:")
print((df['KWH/hh (per half hour)'] == 0).sum())

print("\n5. Consumption categories:")
print(df['Consumption_Category'].value_counts())

print("\n6. Tariff categories:")
print(df['stdorToU'].value_counts())

print("\n7. Dataset information:")
print(df.info())
```

```python

energy = 'KWH/hh (per half hour)'

# Outlier detection using IQR
Q1 = df[energy].quantile(0.25)
Q3 = df[energy].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

outliers = df[df[energy] > upper_limit]

print("Potential unusual readings:", len(outliers))

# Missing half-hour intervals
time_diff = df.groupby('LCLid')['DateTime'].diff()

gaps = time_diff > pd.Timedelta(minutes=30)

print("Intervals with gaps:", gaps.sum())

# Zero readings
print("Zero consumption:", (df[energy] == 0).sum())
```
